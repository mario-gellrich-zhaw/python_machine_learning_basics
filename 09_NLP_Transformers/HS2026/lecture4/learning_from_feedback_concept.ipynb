# Learning from Feedback - Concepts

**Business problem:** choose a featured ad while learning from uncertain responses.

A model supports a decision; it does not make the business decision for us.

## Learning objectives

- Explain exploration and exploitation.
- Distinguish bandits from action-dependent sequential decisions.
- Distinguish supervised LLM learning from reward-based training.

## Dataset and practical scope

All clicks and rates are simulated. The assistant example is hypothetical, not an RL training run.

All code is supplied. Run the cells from top to bottom.

## Libraries and settings

Use the prepared Python 3.11 Codespace kernel. No GPU or API key is needed.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

SEED = 7
np.random.seed(SEED)
DATA_ROOT = Path('../../../00_Data')
```

## 1. Baseline: only the displayed ad gives feedback

An arm is an available choice. A reward is the specified response, here a click. A policy is the choice rule. Fixed A and random selection are baselines, not complete A/B experiments.

```python
display(pd.DataFrame({'ad': ['A', 'B'], 'observed clicks': [2, 1], 'displays': [10, 10]})
        .assign(observed_rate=lambda frame: frame['observed clicks'] / frame['displays']))
```

## 2. Usually use, sometimes try

Epsilon-greedy explores randomly on about 10% of decisions and otherwise uses the highest estimated rate. Exploration can choose the leader too; it does not mean 10% exposure to the weaker ad.

```python
rng = np.random.default_rng(SEED)
estimates = np.array([0.2, 0.1])
explore = rng.random() < 0.1
ad = int(rng.integers(2)) if explore else int(rng.choice(np.flatnonzero(estimates == estimates.max())))
print('Explore?', explore, '| Chosen ad:', ['A', 'B'][ad])
```

## 3. Bandits versus sequential reinforcement learning

Both can seek accumulated reward. Our bandit updates estimates without modelling an action-controlled visitor state. Sequential RL models how a choice changes later situations. Asking for a missing location preference costs an interaction but can improve a later recommendation. A scripted assistant is not automatically RL-trained.

## 4. How human feedback connects to LLMs

1. Text supplies next-token targets; example answers supply supervised fine-tuning targets.
2. People compare answers; a separate scoring model learns those comparisons with supervised learning.
3. Reward-based training can then update the language-model policy.

RLHF means reinforcement learning from human feedback. This classic pathway is not universal or epsilon-greedy. Prompting normally does not update weights.

## Evaluation and interpretation

More simulated clicks are not evidence of better real job matches. A bandit can optimize a poor proxy, and human preferences do not guarantee truth.

## Summary and business implications

Trying reveals missing feedback. Average success can hide individual failures. Always ask whether the reward measures useful business outcomes.

## Limitations

This is a small teaching example, not evidence of deployment performance. Keep evaluation separate from fitting and model selection. Check errors and the user's decision, not only a headline metric.
